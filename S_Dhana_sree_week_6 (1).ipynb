```python
df = df.drop_duplicates()
```

```python
df["Date of Admission"] = pd.to_datetime(df["Date of Admission"])
```

```python
df["Discharge Date"] = pd.to_datetime(df["Discharge Date"])
```

```python
df.info()
```

```python
#Gender distribution

sns.countplot(data = df, x = "Gender")
plt.title("Patient Distribution by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Patients")
plt.show()
```

```python
#Age group distribution

age_bins = [-1, 17, 39, 59, float("inf")]

age_labels = [
    "Child",
    "Young Adults",
    "Middle Adults",
    "Senior "
]

df["Age Group"] = pd.cut(
    df["Age"],
    bins = age_bins,
    labels = age_labels
)
```

```python
sns.scatterplot(data = df, x = "Age", y = "Billing Amount")
plt.title("Age vs Billing Amount")
plt.xlabel("Age")
plt.ylabel("Billing Amount")
plt.show()
```

```python
df["Age Group"].value_counts()
```

```python
sns.countplot(data = df, x = "Age Group")
plt.title("Patient Distribution by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Number of Patients")
plt.show()
```

```python
#Understand Admission

df["Date of Admission"].head()
```

```python
df["Date of Admission"].dtype
```

```python
df["Admission year"] = df["Date of Admission"].dt.year
```

```python
df["Admission Month"] = df["Date of Admission"].dt.month
```

```python
df[["Date of Admission", "Admission year", "Admission Month"]].head()
```

```python
#Count Monthly  Admissions

Monthly_admissions = df.groupby("Admission Month").size().reset_index(name = "Admissions")
```

```python
Monthly_admissions
```

```python
#Admission trend

sns.lineplot(data = Monthly_admissions, x = "Admission Month", y = "Admissions",marker="o")

plt.title("Monthly Admission Trend")
plt.xlabel("Month")
plt.ylabel("Number of Admissions")
plt.xticks(rotation=45)

plt.show()
```

```python
#Compare admission types over time

monthly_admission_types = (
    df.groupby(
        ["Admission Month", "Admission Type"]
        )
    .size()
    .reset_index(name = "Admissions")
)
```

```python
monthly_admission_types.head()
```

```python
sns.lineplot(
    data=monthly_admission_types,
    x="Admission Month",
    y="Admissions",
    hue="Admission Type",
    marker="o"
)
```

```python
df.info()
```

```python
df["Length of Stay"] = (df["Discharge Date"] - df["Date of Admission"]).dt.days
```

```python
sns.scatterplot(data = df, x = "Length of Stay", y = "Billing Amount")

plt.title("Length of Stay vs Billing Amount")
plt.xlabel("Length of Stay")
plt.ylabel("Billing Amount")

plt.show()
```

```python
numeric_cols = [
    "Age",
    "Billing Amount",
    "Length of Stay"
]

corr = df[numeric_cols].corr()
corr
```

```python
plt.figure(figsize = (7, 5))
sns.heatmap(corr, annot = True, cmap = "coolwarm",vmin=-1,vmax=1)
plt.title("Healthcare variables correlation matrix")
plt.show()
```
